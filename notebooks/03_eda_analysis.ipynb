```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

plt.style.use("ggplot")

print("Libraries Loaded")
```

```python
fund_master = pd.read_csv("../data/raw/01_fund_master.csv")
nav_history = pd.read_csv("../data/raw/02_nav_history.csv")
aum = pd.read_csv("../data/raw/03_aum_by_fund_house.csv")
sip = pd.read_csv("../data/raw/04_monthly_sip_inflows.csv")
category = pd.read_csv("../data/raw/05_category_inflows.csv")
folio = pd.read_csv("../data/raw/06_industry_folio_count.csv")
performance = pd.read_csv("../data/raw/07_scheme_performance.csv")
transactions = pd.read_csv("../data/raw/08_investor_transactions.csv")
holdings = pd.read_csv("../data/raw/09_portfolio_holdings.csv")
benchmark = pd.read_csv("../data/raw/10_benchmark_indices.csv")

print("All datasets loaded successfully")
```

```python
print("Fund Master:", fund_master.shape)
print("NAV History:", nav_history.shape)
print("AUM:", aum.shape)
print("SIP:", sip.shape)
print("Transactions:", transactions.shape)
```

```python
fund_master["fund_house"].value_counts().plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Fund House Distribution")
plt.ylabel("Number of Schemes")
plt.show()
```

```python
fund_master["category"].value_counts().plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Fund Category Distribution")
plt.ylabel("Count")
plt.show()
```

```python
sns.countplot(
    data=fund_master,
    x="risk_category"
)

plt.xticks(rotation=45)
plt.title("Risk Category Distribution")
plt.show()
```

```python
top_aum = performance.sort_values(
    by="aum_crore",
    ascending=False
).head(10)

plt.figure(figsize=(10,5))

sns.barplot(
    data=top_aum,
    x="aum_crore",
    y="scheme_name"
)

plt.title("Top 10 Funds by AUM")
plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.plot(
    sip["month"],
    sip["sip_inflow_crore"]
)

plt.xticks(rotation=90)

plt.title("Monthly SIP Inflows Trend")
plt.ylabel("SIP Inflow (Crore)")
plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.plot(
    folio["month"],
    folio["total_folios_crore"]
)

plt.xticks(rotation=90)

plt.title("Industry Folio Growth")
plt.ylabel("Folios (Crore)")
plt.show()
```

```python
transactions["transaction_type"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Transaction Type Distribution")
plt.ylabel("")
plt.show()
```

```python
transactions["state"].value_counts().head(10).plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Top States by Transactions")
plt.ylabel("Count")
plt.show()
```

```python
sns.countplot(
    data=transactions,
    x="age_group"
)

plt.title("Investor Age Groups")
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.hist(
    transactions["annual_income_lakh"],
    bins=20
)

plt.title("Investor Income Distribution")
plt.xlabel("Income (Lakh)")
plt.ylabel("Count")
plt.show()
```

```python
category.groupby("category")["net_inflow_crore"].sum().sort_values(
    ascending=False
).plot(
    kind="bar",
    figsize=(12,5)
)

plt.title("Total Net Inflows by Category")
plt.ylabel("Crore")
plt.show()
```

```python
fund_master["fund_manager"].value_counts().head(10).plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Top Fund Managers by Number of Schemes")
plt.ylabel("Schemes Managed")
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.hist(
    fund_master["expense_ratio_pct"],
    bins=15
)

plt.title("Expense Ratio Distribution")
plt.xlabel("Expense Ratio (%)")
plt.ylabel("Count")
plt.show()
```

```python
benchmark["index_name"].value_counts().plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Benchmark Index Records")
plt.ylabel("Count")
plt.show()
```

```python
holdings["sector"].value_counts().head(10).plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Top Sectors in Portfolio Holdings")
plt.ylabel("Count")
plt.show()
```

```python
# Key Insights

1. SBI, HDFC and ICICI manage a large number of schemes.
2. Equity-oriented categories dominate the mutual fund industry.
3. Moderate and High risk schemes are the most common.
4. SIP inflows show consistent growth over time.
5. Industry folio counts have increased significantly since 2022.
6. SIP transactions dominate investor activity.
7. Maharashtra and Karnataka contribute a large share of transactions.
8. Investors aged 26–45 form the largest investor segment.
9. Large Cap, Flexi Cap and Mid Cap categories attract major inflows.
10. Banking, Pharma and Technology sectors have significant portfolio exposure.
```
